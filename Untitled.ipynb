```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans

df = pd.read_excel("Book1.xlsx")

# --- Minimum minutes filter (removes small-sample noise) ---
MIN_MINUTES = 900
df = df[df["Minutes played"] >= MIN_MINUTES].reset_index(drop=True)

# --- Convert counting stats to per-90 ---
count_features = [
    "Tackles", "Interceptions", "Clearances", "Blocked shots",
    "Errors leading to goal", "Errors leading to shot", "Dribbled past",
    "Aerial duels won", "Fouls", "Dispossessed",
    "Accurate passes", "Long balls (accurate)", "Key passes", "Clean sheets"
]
per90 = df["Minutes played"] / 90
for col in count_features:
    df[col + "_p90"] = df[col] / per90

# --- Rate stats used as-is ---
rate_features = ["Aerial duels won %", "Accurate passes %", "Accurate long balls %"]

feature_cols = [c + "_p90" for c in count_features] + rate_features
data = df[feature_cols].fillna(0)

# --- Standardize + PCA ---
scaler = StandardScaler()
X = scaler.fit_transform(data)

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)
df["PC1"], df["PC2"] = X_pca[:, 0], X_pca[:, 1]

loadings = pd.DataFrame(pca.components_, columns=feature_cols, index=["PC1", "PC2"]).T
print(loadings.sort_values("PC1", ascending=False))
print("Explained variance:", pca.explained_variance_ratio_)

# --- Cluster ---
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
df["Cluster"] = kmeans.fit_predict(X)
print(df["Cluster"].value_counts())

df.to_csv("cb_clustered_v2.csv", index=False)
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.metrics.pairwise import cosine_similarity


# ---------------------------------------------------------------------------
# Build the per-90 feature matrix (same as the archetype model)
# ---------------------------------------------------------------------------
def build_feature_matrix(df, min_minutes=900):
    df = df[df["Minutes played"] >= min_minutes].reset_index(drop=True).copy()

    count_features = [
        "Tackles", "Interceptions", "Clearances", "Blocked shots",
        "Errors leading to goal", "Errors leading to shot", "Dribbled past",
        "Aerial duels won", "Fouls", "Dispossessed",
        "Accurate passes", "Long balls (accurate)", "Key passes", "Clean sheets"
    ]
    per90 = df["Minutes played"] / 90
    for col in count_features:
        df[col + "_p90"] = df[col] / per90

    rate_features = ["Aerial duels won %", "Accurate passes %", "Accurate long balls %"]
    feature_cols = [c + "_p90" for c in count_features] + rate_features

    scaler = StandardScaler()
    X = scaler.fit_transform(df[feature_cols].fillna(0))

    return df, X, feature_cols


# ---------------------------------------------------------------------------
# Objective 1: Tactical fit — cosine similarity, no training needed
# ---------------------------------------------------------------------------
def tactical_fit_score(df, X, target_name):
    idx = df.index[df["Name"] == target_name]
    if len(idx) == 0:
        raise ValueError(f"'{target_name}' not found in dataset")
    idx = idx[0]

    sims = cosine_similarity(X[idx].reshape(1, -1), X).flatten()
    return pd.Series((sims + 1) / 2, index=df.index)  # rescale -1..1 -> 0..1


# ---------------------------------------------------------------------------
# Objective 2: Budget fit — deterministic formula
# ---------------------------------------------------------------------------
def budget_fit_score(df, budget_eur_m):
    value = df["Market Value (€m)"]
    score = np.where(
        value <= budget_eur_m,
        1.0,
        np.maximum(0, 1 - (value - budget_eur_m) / budget_eur_m)
    )
    return pd.Series(score, index=df.index)


# ---------------------------------------------------------------------------
# Objective 3: Age fit — deterministic Gaussian falloff
# ---------------------------------------------------------------------------
def age_fit_score(df, target_age, tolerance=4):
    age = df["Age"]
    score = np.exp(-((age - target_age) ** 2) / (2 * tolerance ** 2))
    return pd.Series(score, index=df.index)


# ---------------------------------------------------------------------------
# Combine into a ranked shortlist
# ---------------------------------------------------------------------------
def rank_candidates(
    df, X,
    target_style_player,
    budget_eur_m,
    target_age,
    weights=None,
    exclude_target=True,
    top_n=10,
):
    if weights is None:
        weights = {"tactical": 0.5, "budget": 0.25, "age": 0.25}
    assert abs(sum(weights.values()) - 1.0) < 1e-6, "weights must sum to 1.0"

    tactical = tactical_fit_score(df, X, target_style_player)
    budget = budget_fit_score(df, budget_eur_m)
    age = age_fit_score(df, target_age)

    blended = (
        weights["tactical"] * tactical +
        weights["budget"] * budget +
        weights["age"] * age
    )

    out = df[["Name", "Team", "Age", "Market Value (€m)", "Contract Expiry"]].copy()
    out["Tactical fit"] = tactical.round(3)
    out["Budget fit"] = budget.round(3)
    out["Age fit"] = age.round(3)
    out["Overall fit"] = blended.round(3)

    if exclude_target:
        out = out[out["Name"] != target_style_player]

    return out.sort_values("Overall fit", ascending=False).head(top_n).reset_index(drop=True)


# ---------------------------------------------------------------------------
# Example usage
# ---------------------------------------------------------------------------
if __name__ == "__main__":
    raw = pd.read_excel("Book1.xlsx")
    df, X, feature_cols = build_feature_matrix(raw)

    results = rank_candidates(
        df, X,
        target_style_player="Jannik Vestergaard",
        budget_eur_m=4.0,
        target_age=25,
        weights={"tactical": 0.5, "budget": 0.3, "age": 0.2},
        top_n=10,
    )
    print(results.to_string(index=False))
```
